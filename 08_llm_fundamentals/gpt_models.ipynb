```python
# Generating Text
def generate_text(prompt):
    response = openai.Completion.create(
        engine= "gpt-4o-mini", # use "gpt-4o-mini" or "davinci-002" for cheapest. "gpt-3.5-turbo" does not work. 
        prompt=prompt,
        max_tokens=10,
        temperature=0.7)
    return response.choices[0].text.strip()
```

```python
prompt = "Once upon a time"
generated_text = generate_text(prompt)
print(prompt, generated_text)
```

```python
# Customizing the output
def generate_text(prompt, max_tokens, temperature):
    response = openai.Completion.create(
        engine="gpt-4o-mini",
        prompt=prompt,
        max_tokens=max_tokens,
        temperature=temperature)
    return response.choices[0].text.strip()
```

```python
# Increase max_tokens Enough to Cover Full Thoughts. For longer sentences or paragraphs, 50 tokens is often too small.
# Try max_tokens=100 or 150 for storytelling-like content.
# This allows the model to reach natural sentence breaks instead of cutting mid-sentence.

generated_text = generate_text(prompt, 120, 0.2)
# print(prompt, generated_text)

# the best way is to remove the incomplete sentences by looking for puncutation
# function for post-processing
import re

def complete_sentences(text):
    sentences = re.findall(r'.*?[.!?]', text)
    return ''.join(sentences)

generated_text = complete_sentences(generated_text)
print(prompt, generated_text)
```
