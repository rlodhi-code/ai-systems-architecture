```python
import openai
# following code is not recommended because config.py file can be eccidentally exposed to GitHub and that exposes the api key.
# use config.py only for saving other configs and not for saving secret keys such as api
# import config
# api_key = config.api_key 
# openai.api_key = api_key
# you can save following in config.py
# import os
# OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

# instead use .env file. Save the key in that file and ensure there is a line .env in .gitignore
# following code is needed
import os
from dotenv import load_dotenv

load_dotenv()

api_key = os.getenv("OPENAI_API_KEY")
openai.api_key = api_key
```

```python
# Generating Text
def generate_text(prompt):
    response = openai.Completion.create(
        engine= "davinci-002", # use "gpt-4o-mini" or "davinci-002" for cheapest. "gpt-3.5-turbo" does not work. 
        prompt=prompt,
        max_tokens=10,
        temperature=0.7)
    return response.choices[0].text.strip()
```

```python
prompt = "Once upon a time"
generated_text = generate_text(prompt)
print(prompt, generated_text)
```

```python
# Customizing the output
def generate_text(prompt, max_tokens, temperature):
    response = openai.Completion.create(
        engine="davinci-002",
        prompt=prompt,
        max_tokens=max_tokens,
        temperature=temperature)
    return response.choices[0].text.strip()
```

```python
generated_text = generate_text(prompt, 50, 0)
print(prompt, generated_text)
```
