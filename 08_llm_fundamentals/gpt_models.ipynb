```python
from openai import OpenAI
# use .env file. Save the key in that file and ensure there is a line .env in .gitignore
# following code is not recommended because config.py file can be accidentally exposed to GitHub and that exposes the api key.
# use config.py only for saving other configs and not for saving secret keys such as api
import os
from dotenv import load_dotenv
load_dotenv()

import config

if config.OPENAI_API_KEY:
    print("API key loaded successfully")
else:
    print("API key not found")

client = OpenAI(api_key=config.OPENAI_API_KEY)
```

```python
# Generating Text
def generate_text(prompt):
    response = client.responses.create(
        model="gpt-4.1-mini",  # modern, fast, cost-effective older ones are "gpt-4o-mini", "davinci-002" or "gpt-3.5-turbo" does not work. 
        input=prompt,
        max_output_tokens=16,
        temperature=0.7,
    )
    return response.output_text.strip()
```

```python
prompt = "Once upon a time"
generated_text = generate_text(prompt)
print(prompt, generated_text)
```

```python
# Customizing the output
def generate_text(prompt, max_tokens, temperature):
    response = client.responses.create(
        model="gpt-4.1-mini",
        input=prompt,
        max_output_tokens=max(max_tokens, 16),  # Responses API requires >= 16
        temperature=temperature,
    )
    return response.output_text.strip()
```

```python
generated_text = generate_text(prompt, 50, 0)
print(prompt, generated_text)
```

```python
# Increase max_tokens Enough to Cover Full Thoughts. For longer sentences or paragraphs, 50 tokens is often too small.
# Try max_tokens=100 or 150 for storytelling-like content.
# This allows the model to reach natural sentence breaks instead of cutting mid-sentence.

generated_text = generate_text(prompt, 120, 0.2)
# print(prompt, generated_text)

# the best way is to remove the incomplete sentences by looking for puncutation
# function for post-processing
import re

def complete_sentences(text):
    sentences = re.findall(r'.*?[.!?]', text)
    return ''.join(sentences)

generated_text = complete_sentences(generated_text)
print(prompt, generated_text)
```

```python
# **🌊 Poetic Chatbot**

def poetic_chatbot(prompt):
    response = client.responses.create(
        model="gpt-4.1-mini",
        input = [
            {
                "role": "system",
                "content": "You are a poetic chatbot."
            },
            {
                "role": "user",
                "content": "When was Google founded?"
            },
            {
                "role": "assistant",
                "content": "In the late '90s, a spark did ignite, Google emerged, a radiant light. By Larry and Sergey, in '98, it was born, a search engine new, on the web it was sworn."
            },
            {
                "role": "user",
                "content": "Which country has the youngest president?"
            },
            {
                "role": "assistant",
                "content": "Ah, the pursuit of youth in politics, a theme we explore. In Austria, Sebastian Kurz did implore, at the age of 31, his journey did begin, leading with vigor, in a world filled with din."
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature = 1,
        
    )
    return response.output_text.strip()
```

```python
prompt = "When was cheese first made?"
poetic_chatbot(prompt)
```
