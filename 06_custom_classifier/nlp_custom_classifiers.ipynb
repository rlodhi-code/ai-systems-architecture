```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
```

```python
data = pd.DataFrame([
    # Positive sentences
    ("Spending a lazy Sunday with coffee and a good book makes me so happy", "positive"),
    ("I just nailed that presentation at work – feels amazing!", "positive"),
    ("My best friend called out of the blue and it instantly brightened my day", "positive"),
    ("The smell of fresh rain always lifts my spirits", "positive"),
    ("I finished my workout and I feel strong and energized", "positive"),
    ("Winning tickets to the concert was such a lucky surprise", "positive"),
    ("I'm so thankful for my supportive family – they mean everything", "positive"),
    ("Finding an old photo from a great trip brought back wonderful memories", "positive"),
    ("I treated myself to my favorite dessert and it was pure bliss", "positive"),
    ("The sunrise this morning was breathtaking and set a perfect tone for the day", "positive"),

    # Negative sentences
    ("I spilled coffee all over my laptop this morning – what a disaster", "negative"),
    ("The meeting dragged on forever and accomplished nothing", "negative"),
    ("I keep getting ignored in group chats and it hurts", "negative"),
    ("My flight got delayed again and now I'll miss the event", "negative"),
    ("I tried a new recipe and it turned out completely inedible", "negative"),
    ("Dealing with rude customers all day drained my energy", "negative"),
    ("I lost my favorite earrings somewhere – I'm gutted", "negative"),
    ("The weather ruined our picnic plans at the last minute", "negative"),
    ("I'm stuck in endless traffic and running late for everything", "negative"),
    ("Hearing bad news about a friend left me feeling really down", "negative")
],
columns=['text', 'sentiment'])
```

```python
data = data.sample(frac=1).reset_index(drop=True)

X = data['text']
y = data['sentiment']

# text vectorization to bow - CountVectorizer
countvec = CountVectorizer()
countvec_fit = countvec.fit_transform(X)
bag_of_words = pd.DataFrame(countvec_fit.toarray(), columns = countvec.get_feature_names_out())

# display the content
bag_of_words
```

```python
# split into train and test data
X_train, X_test, y_train, y_test = train_test_split(bag_of_words, y, test_size=0.3, random_state = 7)
```

```python
# Logistic Regression
lr = LogisticRegression(random_state=1).fit(X_train, y_train)
y_pred_lr = lr.predict(X_test)
accuracy_score(y_pred_lr, y_test)
```

```python
y_pred_lr
```

```python
y_test
```

```python
print(classification_report(y_test, y_pred_lr, zero_division=0))
```

```python
# Naive Bayes
from sklearn.naive_bayes import MultinomialNB
nb = MultinomialNB().fit(X_train, y_train)
y_pred_nb = nb.predict(X_test)
accuracy_score(y_pred_nb, y_test)
```

```python
print(classification_report(y_test, y_pred_nb, zero_division=0))
```

```python
# Linear Support Vector Machine
from sklearn.linear_model import LogisticRegression, SGDClassifier
svm = SGDClassifier().fit(X_train, y_train)
# possible hyper params, loss function, regularization
y_pred_svm = svm.predict(X_test)
accuracy_score(y_pred_svm, y_test)
```

```python
print(classification_report(y_test, y_pred_svm, zero_division=0))
```
