```python
from textblob import TextBlob
```

```python
sentence_1 = "i had a great time at the movie it was really funny"
sentence_2 = "i had a great time at the movie but the parking was terrible"
sentence_3 = "i had a great time at the movie but the parking wasn't great"
sentence_4 = "i went to see a movie"
```

```python
# Lexicon. TextBlob returns a score between -1 and 1, where negative scores indicate negative sentiment, and positive scores indicate positive sentiment.
print(sentence_1)
sentiment_score = TextBlob(sentence_1)
print(sentiment_score.sentiment.polarity)

print(sentence_2)
sentiment_score_2 = TextBlob(sentence_2)
print(sentiment_score_2.sentiment.polarity)

print(sentence_3)
sentiment_score_3 = TextBlob(sentence_3)
print(sentiment_score_3.sentiment.polarity)

print(sentence_4)
sentiment_score_4 = TextBlob(sentence_4)
print(sentiment_score_4.sentiment.polarity)
```

```python
# VADER (Valence Aware Dictionary and sEntiment Reasoner)
""" improves classical lexicons by adding:
Intensity handling
Negation detection
Punctuation emphasis
Capitalization emphasis
Degree modifiers (“very”, “extremely”)
Emoji & slang support
"""
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
```

```python
vader_sentiment = SentimentIntensityAnalyzer()
print(sentence_1)
print(vader_sentiment.polarity_scores(sentence_1))
print(sentence_2)
print(vader_sentiment.polarity_scores(sentence_2)) 
print(sentence_3)
print(vader_sentiment.polarity_scores(sentence_3)) 
print(sentence_4)
print(vader_sentiment.polarity_scores(sentence_4)) 
```

```python
# pre-trained models. The score gives the model confidence level. Following code only provides positive and negative sentiments. The sentence 2 
# shows up NEG with 0.997 score.  That was misclassified. That needs to be addressed. 
import transformers
from transformers import pipeline
# import emoji for the warning messages
import emoji
print(emoji.__version__)

sentiment_pipeline = pipeline("sentiment-analysis")
print(sentence_1)
print(sentiment_pipeline(sentence_1))
print(sentence_2)
print(sentiment_pipeline(sentence_2))
print(sentence_3)
print(sentiment_pipeline(sentence_3))
print(sentence_4)
print(sentiment_pipeline(sentence_4))
```
